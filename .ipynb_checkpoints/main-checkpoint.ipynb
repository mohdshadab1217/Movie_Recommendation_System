## Content Based Movie Recommendation
- The dataset here is used to train this model is taken from Kaggle's website.

```python
#function to train model
import pandas as pd
import re
from sklearn.feature_extraction.text import TfidfVectorizer,ENGLISH_STOP_WORDS
from sklearn.neighbors import NearestNeighbors
def train():
    global model,tv,df
    
    #loading dataset
    df=pd.read_csv('dataset\movies_content.csv',usecols=['description','language','rating','cast','genre','name']).dropna()
    df.reset_index(drop=True,inplace=True)
    
    #cleaning dataset (substitute !@#$)%(+=,.?";:' by "" and - by " " )
    corpus=[]
    for str in df.description:
        sentence=[]
        for word in str.split():
            word=re.sub(r'''[!@#$)%(+=,|.?";:']''',"",word)
            if '-' in list(word):
                word=re.sub('[-]',' ',word)
            sentence.append(word)
        sentence=" ".join(sentence)
        corpus.append(sentence)
    
    #extracting features
    tv=TfidfVectorizer(lowercase=True,stop_words=list(ENGLISH_STOP_WORDS))
    X=tv.fit_transform(corpus).toarray()
    
    #train model
    model=NearestNeighbors(metric='cosine',n_neighbors=6)
    model.fit(X)
train()
```

```python
# find similar by providing name of movie. The name should be exactly same as in dataset.
name=df.name[88]
name
```

```python
#find similar movie
story=df[df.name==name].description
corpus=[]
for str in story:
    sentence=[]
    for word in str.split():
        word=re.sub(r'''[!@#$)%(+=,.?";:']''',"",word)
        if '-' in list(word):
            word=re.sub('[-]',' ',word)
        sentence.append(word)
    sentence=" ".join(sentence)
    corpus.append(sentence)

samples=tv.transform(corpus).toarray()
result=model.kneighbors(samples)
result
```

```python
#print name of movie and 5 similar movies that you got earlier
print(f'Enter Movie Name = {name}')
for i,similar in enumerate(result[1][0][1:],start=1):
    print(f'{i}. Similar Movie Name = {df.name[similar]}')
```
